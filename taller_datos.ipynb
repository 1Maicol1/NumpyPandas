## Ejercicio 1: Operaciones fundamentales con NumPy

```python
import numpy as np

arr_1d = np.array([1, 2, 3, 4, 5])
print("Arreglo 1D:", arr_1d)
print("Forma (shape):", arr_1d.shape)
print("Tipo de dato (dtype):", arr_1d.dtype)
```

```python
arr_2d = arr_1d.reshape(1, 5)
print("Arreglo reshape(1, 5):\n", arr_2d)
print("Forma:", arr_2d.shape)
```

```python
rand_matrix = np.random.rand(3, 3)
int_matrix = np.random.randint(1, 10, size=(3, 3))

print("np.random.rand():\n", rand_matrix)
print("\nnp.random.randint():\n", int_matrix)
```

```python
zeros_matrix = np.zeros((2, 3))
ones_matrix = np.ones((2, 3))
identity_matrix = np.eye(4)

print("np.zeros():\n", zeros_matrix)
print("\nnp.ones():\n", ones_matrix)
print("\nnp.eye():\n", identity_matrix)
```

```python
matriz_aleatoria = np.random.randint(1, 10, size=(2, 4))
print("Matriz original:\n", matriz_aleatoria)
print("\nTranspuesta ( .T ):\n", matriz_aleatoria.T)
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

print("np.hstack():", np.hstack((a, b)))
print("np.vstack():\n", np.vstack((a, b)))
```

```python
arr_stats = np.array([2, 5, 8, 10, 12])

print("Suma:", np.sum(arr_stats))
print("Promedio:", np.mean(arr_stats))
print("Mediana:", np.median(arr_stats))
print("Producto:", np.prod(arr_stats))
print("Desviación estándar:", np.std(arr_stats))
print("Varianza:", np.var(arr_stats))
print("Mínimo:", np.min(arr_stats))
print("Máximo:", np.max(arr_stats))
print("Suma acumulativa:", np.cumsum(arr_stats))
```

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print("Suma elemento a elemento:", x + y)
print("Resta elemento a elemento:", x - y)
print("Multiplicación elemento a elemento:", x * y)
```

## Ejercicio 2: Integración y manipulación con Pandas

```python
import pandas as pd
import numpy as np

print("Pandas y NumPy importados correctamente.")
```

```python
array_2d = np.array([[10, 20, 30], [40, 50, 60]])
df_from_array = pd.DataFrame(array_2d, columns=["A", "B", "C"])
print("DataFrame desde arreglo NumPy:\n", df_from_array)
```

```python
datos_dict = {
    "Nombre": ["Ana", "Luis", "Sofía"],
    "Edad": [22, 25, 29],
    "Ciudad": ["Bogotá", "Medellín", "Cali"]
}
df_from_dict = pd.DataFrame(datos_dict)
print("DataFrame desde diccionario:\n", df_from_dict)
```

```python
arr_from_df = df_from_array.to_numpy()
arr_from_df_values = df_from_array.values

print("Desde .to_numpy():\n", arr_from_df)
print("\nDesde .values:\n", arr_from_df_values)
```

```python
promedio_columna_np = np.mean(df_from_array, axis=0)
promedio_fila_np = np.mean(df_from_array, axis=1)

print("Promedio por columna con np.mean(axis=0):", promedio_columna_np)
print("Promedio por fila con np.mean(axis=1):", promedio_fila_np)
```

```python
promedio_columna_df = df_from_array.mean(axis=0)
promedio_fila_df = df_from_array.mean(axis=1)

print("df.mean(axis=0):\n", promedio_columna_df)
print("\ndf.mean(axis=1):\n", promedio_fila_df)
```
